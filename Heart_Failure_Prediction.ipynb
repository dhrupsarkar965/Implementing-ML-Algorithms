```python
x=df.iloc[0:,0:11]
y=df['HeartDisease']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LogisticRegression
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
categorical_cols=x_train[['Sex','ChestPainType','RestingECG','ExerciseAngina','ST_Slope']]
numerical_cols=x_train[['Age','RestingBP','Cholesterol','FastingBS','MaxHR','Oldpeak']]
```

```python

categorical_pipe=Pipeline(steps=[

                        ('categorical',OneHotEncoder())

                         ])
```

```python
numerical_pipe=Pipeline(steps=[
    ('scaler',StandardScaler())
])
```

```python
transformer=ColumnTransformer(transformers=[
    ('Categorical',categorical_pipe,categorical_cols.columns),
    ('Numerical',numerical_pipe,numerical_cols.columns)
],remainder='passthrough')
```

```python
x_trains_trans=transformer.fit_transform(x_train)
x_test_trans=transformer.transform(x_test)
```

```python
Model=LogisticRegression()
Model.fit(x_trains_trans,y_train)
y_pred=Model.predict(x_test_trans)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,recall_score,precision_score,classification_report,f1_score


confusion_matrix(y_test,y_pred)
```

```python
print(accuracy_score(y_test,y_pred))
print(recall_score(y_test,y_pred))
print(precision_score(y_test,y_pred))
print(f1_score(y_test,y_pred))
```

```python
print(classification_report(y_test,y_pred))
```

```python
from sklearn.metrics import roc_auc_score
model = LogisticRegression()
model.fit(x_trains_trans, y_train)
```

```python

y_prob = model.predict_proba(x_test_trans)[:, 1]

roc_auc = roc_auc_score(y_test, y_prob)

print("ROC-AUC Score:", roc_auc)
```
